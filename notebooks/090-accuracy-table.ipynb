```python
from pathlib import Path

import pandas as pd

GENERATED_RESULTS = Path("..") / "generated_results"

# Per-dataset mean test accuracy over resamples, datasets sorted alphabetically, methods in
# the accuracy CD diagram's rank order.
means = pd.read_csv(GENERATED_RESULTS / "tscbench" / "Accuracy" / "accuracy_mean.csv", index_col=0)
means = means.rename(columns={
    "TSCGlue-high-accuracy": "TSCGlue-H",
    "TSCGlue-medium-accuracy": "TSCGlue-M",
    "TSCGlue-low-accuracy": "TSCGlue-L",
    "WEASEL_V2": "WEASEL-V2",
})
methods = [
    "TSCGlue-H", "TSCGlue-M", "TSCGlue-L", "HIVECOTEV2", "MultiRocketHydra", "RDST",
    "WEASEL-V2", "QUANT", "ROCKET", "FreshPRINCE", "Catch22",
]
means = means[methods].dropna().sort_index(key=lambda s: s.str.lower()).rename_axis("Dataset")

means.to_csv(Path("figures2") / "accuracy_per_dataset.csv", float_format="%.3f")
print(means.to_csv(float_format="%.3f"))
```
