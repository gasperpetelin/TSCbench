```python
import tempfile
import time

from aeon.datasets import load_classification
from tscglue.models import TSCGlueEnhancedV4

X_train, y_train = load_classification("Wafer", split="train", extract_path="../data")
X_train.shape
```

```python
with tempfile.TemporaryDirectory() as runs_dir:
    clf = TSCGlueEnhancedV4(verbose=10, random_state=0, n_jobs=1, n_gpus=0, eval_metric="log_loss", preset="medium", runs_dir=runs_dir)
    t0 = time.perf_counter()
    clf.fit(X_train, y_train)
    fit_s = time.perf_counter() - t0
    print(f"1 cpu, 0 gpu: fit {fit_s:.1f}s")
```
