# ROCKET vs Catch22 over the UCR-112

**Step 1 — run:** fit each classifier on every dataset/resample and write tsml-format result files to `OUT`.

```python
import random
from itertools import product

from aeon.datasets.tsc_datasets import univariate_equal_length
from tsml_eval.experiments import load_and_run_classification_experiment
from tsml_eval.publications.y2023.tsc_bakeoff.set_bakeoff_classifier import (
    _set_bakeoff_classifier,
)
from tscglue.models import TSCGlueClassifier
from tqdm.auto import tqdm

DATA = "/home/petelin/TSCbench/data"          # has <ds>/<ds>{r}_TRAIN.ts resample folds
OUT = "./model_results"                       # tsml-format prediction files land here
N_JOBS = 32
CLASSIFIERS = ["ROCKET", "MiniRocket", "Catch22", "TSCGlue-Accuracy", "TSCGlue-LogLoss"]
RESAMPLES = range(1)                          # resample 0 to start; use range(30) for the full bake-off set

datasets = sorted(univariate_equal_length)    # the 112 UCR equal-length datasets


def make_classifier(name, random_state, n_jobs):
    if name == "TSCGlue-Accuracy":
        return TSCGlueClassifier(verbose=10, random_state=random_state, n_jobs=n_jobs, eval_metric="accuracy")
    elif name == "TSCGlue-LogLoss":
        return TSCGlueClassifier(verbose=10, random_state=random_state, n_jobs=n_jobs, eval_metric="log_loss")
    return _set_bakeoff_classifier(name, random_state=random_state, n_jobs=n_jobs)


# Build every (classifier, dataset, resample) combination up front, then run as one loop.
combos = list(product(CLASSIFIERS, datasets, RESAMPLES))
random.shuffle(combos)
print(len(datasets), "datasets ->", len(combos), "runs")

for clf_name, dataset, r in tqdm(combos):
    try:
        tqdm.write(f"{clf_name} {dataset} {r}")
        load_and_run_classification_experiment(
            DATA, OUT, dataset,
            make_classifier(clf_name, random_state=r, n_jobs=N_JOBS),
            classifier_name=clf_name,
            resample_id=r,
            predefined_resample=True,   # use the local resample fold files
            overwrite=False,            # resumable: skip what's already written
        )
    except Exception as exc:
        print(f"ERROR {clf_name} {dataset} resample={r}: {exc}")
```

**Step 2 — summary:** read the result files back and produce per-metric summary CSVs and critical-difference diagrams under `./generated_results/`.

```python
from tsml_eval.evaluation import evaluate_classifiers_by_problem

# Summarize the results generated above (reads OUT, reuses CLASSIFIERS/datasets/RESAMPLES).
evaluate_classifiers_by_problem(
    OUT,
    CLASSIFIERS,
    datasets,
    "./generated_results/",          # summary CSVs + critical-difference diagrams land here
    resamples=len(list(RESAMPLES)),
    #eval_name="rocket_vs_catch22",
)
```
