```python
!uv pip install autogluon.tabular[xgboost,fastai]
import tempfile
from sklearn.metrics import accuracy_score
from tscglue.models import TSCGlueClassifier, TSCAGGlueClassifier
from tscbench.utils import load_ucr_fold
```

```python
DATASET = "ArrowHead"
FOLD = 5

X_train, y_train, X_test, y_test = load_ucr_fold(DATASET, FOLD)
print(f"Train: {X_train.shape}, Test: {X_test.shape}, Classes: {sorted(set(y_test))}")
```

```python
from time import perf_counter

with tempfile.TemporaryDirectory(prefix="tscagglue_run_") as tmp_dir:
    model = TSCAGGlueClassifier(n_jobs=12, runs_dir=tmp_dir, verbose=10, ag_time_limit=5*60)
    model.fit(X_train, y_train)
    y_pred_agg = model.predict(X_test)

acc_agg = accuracy_score(y_test, y_pred_agg)
print(f"TSCAGGlueClassifier  accuracy: {acc_agg:.4f}")
```

```python
with tempfile.TemporaryDirectory(prefix="tscglue_run_") as tmp_dir:
    model = TSCGlueClassifier(n_jobs=12, runs_dir=tmp_dir, verbose=10)
    model.fit(X_train, y_train)
    y_pred_glue = model.predict(X_test)

acc_glue = accuracy_score(y_test, y_pred_glue)
print(f"TSCGlueClassifier    accuracy: {acc_glue:.4f}")
```

```python
print(f"TSCGlueClassifier    accuracy: {acc_glue}")
print(f"TSCAGGlueClassifier  accuracy: {acc_agg}")
print(f"Delta:               {acc_agg - acc_glue}")
```
