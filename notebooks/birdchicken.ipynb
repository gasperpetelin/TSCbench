```python
import matplotlib.pyplot as plt
from tscbench.utils import load_ucr_fold

X_train, y_train, X_test, y_test = load_ucr_fold("BirdChicken", fold=0)
classes = sorted(set(y_train))

fig, axes = plt.subplots(len(classes), 1, figsize=(14, 4 * len(classes)), sharex=True)
for ax, cls in zip(axes, classes):
    idx = [i for i, y in enumerate(y_train) if y == cls]
    for i in idx:
        ax.plot(X_train[i, 0], alpha=0.5, linewidth=0.8)
    ax.set_title(f"Class {cls}  (n={len(idx)})")
    ax.set_ylabel("amplitude")
    ax.grid(alpha=0.3)
axes[-1].set_xlabel("time")
fig.suptitle("BirdChicken — train series by class", fontsize=13)
plt.tight_layout()
plt.show()
```
