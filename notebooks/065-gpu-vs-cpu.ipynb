```python
import os
import time

os.environ["TF_FORCE_GPU_ALLOW_GROWTH"] = "true"

#from tscglue import data_loader

#data_loader.DATA_DIR = "/home/petelin/TSCbench/data"
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, log_loss

from tscbench.utils import load_ucr_fold
from tscglue.models import TSCGlueClassifier
```

```python
dataset = "Worms"
# dataset = 'Car'
dataset = 'HandOutlines'
# dataset = 'Trace'
# dataset = 'SwedishLeaf'
# dataset = 'Meat'
# dataset='ACSF1'
# dataset='ElectricDevices'
X_train, y_train, X_test, y_test = load_ucr_fold(dataset, 19)
print(X_train.shape, y_train.shape, X_test.shape, y_test.shape)
```

```python
seed = 2683
```

## Run without GPU (n_gpus=0)

```python
m_cpu = TSCGlueClassifier(random_state=seed, n_jobs=16, verbose=10, n_gpus=0)

t0 = time.perf_counter()
m_cpu.fit(X_train, y_train)
cpu_fit_time = time.perf_counter() - t0
print(f"\nCPU fit time: {cpu_fit_time:.2f}s")
```

```python
t0 = time.perf_counter()
y_pred_cpu = m_cpu.predict(X_test)
cpu_pred_time = time.perf_counter() - t0

acc_cpu = accuracy_score(y_test, y_pred_cpu)
print(f"CPU predict time: {cpu_pred_time:.2f}s")
print(f"CPU accuracy on {dataset}: {acc_cpu:.4f}")
```

## Run with GPU (n_gpus=1)

```python
m_gpu = TSCGlueClassifier(random_state=seed, n_jobs=16, verbose=10, n_gpus=1)

t0 = time.perf_counter()
m_gpu.fit(X_train, y_train)
gpu_fit_time = time.perf_counter() - t0
print(f"\nGPU fit time: {gpu_fit_time:.2f}s")
```

```python
t0 = time.perf_counter()
y_pred_gpu = m_gpu.predict(X_test)
gpu_pred_time = time.perf_counter() - t0

acc_gpu = accuracy_score(y_test, y_pred_gpu)
print(f"GPU predict time: {gpu_pred_time:.2f}s")
print(f"GPU accuracy on {dataset}: {acc_gpu:.4f}")
```

## Summary

```python
print(f"{'':12s}  {'fit (s)':>10s}  {'predict (s)':>12s}  {'accuracy':>10s}")
print("-" * 52)
print(f"{'CPU (n_gpus=0)':12s}  {cpu_fit_time:>10.2f}  {cpu_pred_time:>12.2f}  {acc_cpu:>10.4f}")
print(f"{'GPU (n_gpus=1)':12s}  {gpu_fit_time:>10.2f}  {gpu_pred_time:>12.2f}  {acc_gpu:>10.4f}")
print()
print(f"GPU speedup (fit):     {cpu_fit_time / gpu_fit_time:.2f}x")
print(f"GPU speedup (predict): {cpu_pred_time / gpu_pred_time:.2f}x")
```

```python
import polars as pl

def transformer_times(model):
    df = pl.DataFrame(model.summary(return_transforms=True))
    return (
        df.filter(pl.col("level").is_null())
        .select("model", "train_time")
        .with_columns(pl.col("train_time").list.first().alias("fit_s"))
        .drop("train_time")
    )

cpu_t = transformer_times(m_cpu).rename({"fit_s": "cpu_fit_s"})
gpu_t = transformer_times(m_gpu).rename({"fit_s": "gpu_fit_s"})

cmp = cpu_t.join(gpu_t, on="model").with_columns(
    (pl.col("cpu_fit_s") / pl.col("gpu_fit_s")).alias("speedup")
).sort("cpu_fit_s", descending=True)

print("Per-transformer timing (fit):")
print(cmp)
```
